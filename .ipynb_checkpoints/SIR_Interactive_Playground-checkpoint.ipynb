# SIR interactive playground

This notebook is an optional visual introduction to SIR dynamics. Use the sliders to see how changing the transmission rate ($\beta$) and recovery rate ($\gamma$) changes the paths of susceptible, infectious, and recovered people over time.

This is an illustrative model with a fixed population of 10,000 and one initial infectious person. These slider values are for exploration; they are not estimates from the testing data. The data-based estimates come later in the main Session 7 notebook.

## How the two rates work together

The basic reproduction number at the start is $R_0=\beta/\gamma$. It compares transmission with recovery:

| Transmission rate $\beta$ | Recovery rate $\gamma$ | Initial pattern in this model |
|---|---|---|
| Low | Low | Both processes are slower. Infections grow if $\beta>\gamma$ and decline if $\beta<\gamma$. |
| Low | High | Recovery is faster relative to transmission; infections initially decline. |
| High | Low | Transmission is faster relative to recovery; infections initially grow. |
| High | High | Both processes are faster. Infections grow if $\beta>\gamma$ and decline if $\beta<\gamma$. |

“High” or “low” by itself does not determine the direction. Compare the rates: $R_0>1$ means initial growth, while $R_0<1$ means initial decline.

## 1. Set the population and time scale

We will track a population of 10,000 people for 60 weekly steps. At time zero, one person is infectious, the rest are susceptible, and nobody has recovered.

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import ipywidgets as widgets
from IPython.display import display

# Enable interactive widgets when this notebook is opened in Google Colab.
try:
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass

N = 10_000
INITIAL_INFECTIOUS = 1
SIMULATION_WEEKS = 60
```

## 2. Choose $\beta$ and $\gamma$

The sliders set the weekly transmission and recovery rates. Try changing one rate at a time, then compare combinations with similar or different $R_0$ values.

```python
beta_slider = widgets.FloatSlider(
    value=0.4, min=0.05, max=0.8, step=0.05,
    description="Beta / week", continuous_update=False,
)
gamma_slider = widgets.FloatSlider(
    value=0.1, min=0.025, max=0.3, step=0.025,
    description="Gamma / week", continuous_update=False,
)
```

## 3. Represent one week of SIR change

Each week, infectious people can transmit to susceptible people, and infectious people can recover. The resulting counts become the starting counts for the next week. Newly infected people are included in the infectious group for the following step.

```python
def simulate_sir(beta, gamma):
    susceptible = N - INITIAL_INFECTIOUS
    infectious = INITIAL_INFECTIOUS
    recovered = 0.0
    rows = [(0, susceptible, infectious, recovered)]

    for week in range(1, SIMULATION_WEEKS + 1):
        new_infections = susceptible * (-np.expm1(-beta * infectious / N))
        new_recoveries = infectious * (-np.expm1(-gamma))
        susceptible -= new_infections
        infectious += new_infections - new_recoveries
        recovered += new_recoveries
        rows.append((week, susceptible, infectious, recovered))

    return pd.DataFrame(rows, columns=["week", "S", "I", "R"])
```

## 4. Explore the changing curves

Move either slider and hover over a week to read the model’s S, I, and R counts. The title also shows $R_0=\beta/\gamma$ for the selected rates.

```python
def show_sir_curves(beta, gamma):
    trajectory = simulate_sir(beta, gamma)
    fig = go.Figure()
    for state, color in [("S", "#087e8b"), ("I", "#e07a32"), ("R", "#7c8790")]:
        fig.add_trace(go.Scatter(
            x=trajectory.week,
            y=trajectory[state],
            mode="lines",
            name=state,
            line={"color": color, "width": 3},
            hovertemplate=f"{state} = %{{y:,.0f}} people<extra></extra>",
        ))

    fig.update_layout(
        title=f"SIR dynamics: beta = {beta:.2f}/week, gamma = {gamma:.3f}/week, R0 = {beta/gamma:.2f}",
        xaxis_title="Weeks after time zero",
        yaxis_title="People",
        hovermode="x unified",
        height=470,
        legend_title="State",
        margin={"l": 60, "r": 25, "t": 75, "b": 55},
    )
    display(fig)

controls = widgets.VBox([beta_slider, gamma_slider])
curves = widgets.interactive_output(
    show_sir_curves, {"beta": beta_slider, "gamma": gamma_slider}
)
display(controls, curves)
```
